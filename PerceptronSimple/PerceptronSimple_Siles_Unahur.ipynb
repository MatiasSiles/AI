```python
import numpy as np
```

```python
# Funciones auxiliares

def calcular_activacion(h):
  """Funcion signo"""
  return 1 if h >= 0 else -1

def calcular_error(X, y, w):
  """Calcula el error absoluto total"""
  error = 0
  for i_X in range(len(X)):
    O = calcular_activacion(X[i_X] @ w)
    error += abs(y[i_X] - O)
  return error
```

```python
def perceptron(X, y, eta, cota):
  """Entrenamiento de perceptron simple"""

  p, n = X.shape
  i = 0
  w = np.zeros(n)
  w[-1] = 1 # bias
  error_min = p * 2
  error = 1

  while error > 0 and i < cota:
    i_X = np.random.randint(0, p)

    h = X[i_X] @ w
    O = calcular_activacion(h)

    delta_w = eta * (y[i_X] - O) * X[i_X]
    w += delta_w

    error = calcular_error(X, y, w)

    if error < error_min:
      error_min = error
      w_min = w.copy()

    i += 1

  return i, error_min, w_min
```

```python
X = np.array([
    [-1,  1],
    [ 1, -1],
    [-1, -1],
    [ 1,  1]
])

y = np.array([-1, -1, -1, 1])

# Agregar bias
bias = np.ones((len(X), 1))
X = np.hstack((X, bias))

# Ejecucion
i, error_min, w_min = perceptron(X, y, eta=0.1, cota=100)

print(f"Iters: {i}")
print(f"Min. error: {error_min}")
print(f"Mejores pesos: {w_min}")
```

# Glosario


*   p = Cantidad de entradas
*   n = Cantidad de dimensiones
*   COTA = Max. de iteraciones
*   i_x = Numero al azar entre 1 y p
*   y = Vector de la salida esperada
*   x = Array de vectores sobre el conjunto de entrenamiento
*   w = Vector de pesos
*   h = Excitacion
*   O = La salida de una entrada al azar en el perceptron
*   eta = Tasa de aprendizaje
*   pos = Posicion
*   Error = 1, Error inicial para entrenar el perceptron
*   arr = array
